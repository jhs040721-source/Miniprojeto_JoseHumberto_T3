```python
#Sprint 2 – Diagnóstico e ajuste dos tipos de dados
#Atividades:
#Identificar os tipos atuais das colunas.
#Padronizar os nomes das colunas.
#Remover espaços desnecessários dos textos.
#Verificar categorias vazias ou escritas de maneiras diferentes.
#Converter a coluna de data para datetime.
#Converter colunas numéricas que estejam armazenadas como texto.
#Identificar datas inválidas geradas durante a conversão.
#Conferir os tipos depois dos ajustes.
# Mostra o tipo atual de cada coluna antes das transformações
tipos_antes = df.dtypes.to_frame(name="Tipo atual")

print("TIPOS DE DADOS ANTES DOS AJUSTES")
display(tipos_antes)
```

```python
# Padroniza os nomes das colunas:
# remove espaços nas extremidades, converte para maiúsculas,
# substitui espaços por "_" e remove o caractere ":"

df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_", regex=False)
    .str.replace(":", "", regex=False)
)

# Mostra os nomes depois da padronização
print("COLUNAS APÓS A PADRONIZAÇÃO:")
print(df.columns.tolist())
```

```python
# Colunas que contêm textos ou categorias
colunas_texto = ["CL_GENERO", "CL_SEG", "PR_CAT", "PR_NOME"]

# Remove espaços desnecessários e converte os textos para maiúsculas
for coluna in colunas_texto:
    df[coluna] = df[coluna].str.strip().str.upper()

print("Textos padronizados com sucesso!")
```

```python
# Verifica a quantidade de categorias, valores nulos e textos vazios
for coluna in colunas_texto:
    print("\nCOLUNA:", coluna)
    print("Categorias diferentes:", df[coluna].nunique())
    print("Valores nulos:", df[coluna].isnull().sum())
    print("Textos vazios:", (df[coluna] == "").sum())

    print("Valores mais frequentes:")
    print(df[coluna].value_counts(dropna=False).head(10))
    print("-" * 50)
```

A verificação encontrou uma inconsistência relevante:

CL_GENERO: 2 categorias válidas (F e M);

não existem textos vazios nem nulos reconhecidos pelo pandas nas colunas textuais.

O valor #N/D ainda não aparece como nulo porque está armazenado como texto. Vamos tratá-lo na Sprint 3.

```python
# Mostra uma amostra dos valores originais da coluna DATA
print("EXEMPLOS DE DATAS DA BASE:")
print(df["DATA"].head(10).tolist())
```

```python
# Guarda uma cópia das datas originais para conferência
datas_originais = df["DATA"].copy()

# Converte a coluna DATA de texto para o tipo datetime
# O formato %d/%m/%Y representa dia/mês/ano
# errors="coerce" transforma datas inválidas em NaT
df["DATA"] = pd.to_datetime(
    df["DATA"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Mostra o novo tipo da coluna e alguns exemplos
print("Novo tipo da coluna DATA:", df["DATA"].dtype)
print("\nPrimeiras datas convertidas:")
print(df["DATA"].head())
```

```python
# Conta quantas datas se tornaram NaT durante a conversão
quantidade_datas_invalidas = df["DATA"].isna().sum()

print("Quantidade de datas inválidas:", quantidade_datas_invalidas)

# Se existirem datas inválidas, mostra os valores originais
if quantidade_datas_invalidas > 0:
    print("\nExemplos de valores inválidos:")
    print(
        datas_originais[df["DATA"].isna()]
        .value_counts()
        .head(10)
    )
else:
    print("Nenhuma data inválida foi encontrada.")
```

### Conversão e validação das datas

A coluna DATA estava armazenada como texto e foi convertida para o tipo datetime, utilizando o formato dia/mês/ano. A conversão não identificou datas inválidas, pois nenhum registro foi transformado em NaT.

```python
# Colunas que devem possuir valores numéricos
colunas_numericas = ["CO_ID", "CL_ID", "CL_EC", "CL_FHL", "PR_ID"]

# Garante que essas colunas estejam armazenadas como números
for coluna in colunas_numericas:
    df[coluna] = pd.to_numeric(df[coluna], errors="coerce")

# Mostra os tipos após a verificação
print("TIPOS DAS COLUNAS NUMÉRICAS:")
print(df[colunas_numericas].dtypes)

# Verifica se a conversão gerou valores nulos
print("\nNULOS NAS COLUNAS NUMÉRICAS:")
print(df[colunas_numericas].isnull().sum())
```

```python
#Sprint 3 – Tratamento de nulos, duplicatas e inconsistências
#Atividades:
#Contar os valores nulos de cada coluna.
#Identificar registros duplicados.
#Verificar categorias vazias ou inconsistentes.
#Definir se os valores nulos serão removidos ou preenchidos.
#Explicar a escolha realizada para tratar os valores nulos.
#Eliminar duplicatas relevantes.
#Conferir o resultado da limpeza.
#Criar o DataFrame limpo.
#Resultado esperado:
#Base sem duplicatas relevantes, com os valores nulos tratados e com as decisões de limpeza explicadas no notebook.

# Conta a quantidade de valores nulos em cada coluna
nulos_por_coluna = df.isnull().sum()

# Calcula o percentual de valores nulos
percentual_nulos = (df.isnull().sum() / len(df) * 100).round(2)

# Reúne os resultados em uma tabela
relatorio_nulos = pd.DataFrame({
    "Quantidade de nulos": nulos_por_coluna,
    "Percentual de nulos": percentual_nulos
})

print("VALORES NULOS POR COLUNA:")
display(relatorio_nulos)
```

```python
# Conta registros completamente duplicados
quantidade_duplicatas = df.duplicated().sum()

print("Quantidade de registros duplicados:", quantidade_duplicatas)

# Mostra alguns exemplos, caso existam duplicatas
if quantidade_duplicatas > 0:
    print("\nExemplos de registros duplicados:")
    display(df[df.duplicated(keep=False)].head(10))
else:
    print("Nenhum registro completamente duplicado foi encontrado.")
```

### Diagnóstico das duplicatas

Foram encontrados 96.553 registros completamente duplicados. Para fins deste mini-projeto, os registros integralmente idênticos serão considerados duplicatas relevantes e removidos.

A base não possui identificador da transação, número da nota fiscal ou quantidade comprada. Por isso, existe a possibilidade de algumas repetições representarem compras legítimas de mais de uma unidade do mesmo produto. Essa limitação será registrada nas conclusões do projeto.

```python
# Verifica textos vazios e valores usados para representar ausência de informação
for coluna in colunas_texto:
    quantidade_vazios = (df[coluna] == "").sum()
    quantidade_nd = (df[coluna] == "#N/D").sum()

    print(
        coluna,
        "| Textos vazios:", quantidade_vazios,
        "| Valores #N/D:", quantidade_nd
    )
```

### Decisão sobre o tratamento dos valores ausentes

Foram adotadas as seguintes decisões:

1. As colunas UNNAMED_10, UNNAMED_11, UNNAMED_12 e UNNAMED_13 serão removidas, pois estão 100% vazias e não possuem informação útil.

2. Os 3.650 valores #N/D encontrados na coluna PR_CAT serão substituídos pela categoria NÃO INFORMADO.

3. Optou-se por não excluir as linhas com #N/D porque as demais informações desses registros, como data, cliente, produto e segmento, continuam úteis para a análise.

4. Não será atribuída uma categoria específica aos registros #N/D, pois isso exigiria uma suposição sobre os dados. A expressão NÃO INFORMADO preserva os registros sem criar uma informação que não estava na base original.

5. As demais colunas principais não apresentaram valores nulos e, portanto, não precisam de remoção ou preenchimento.

```python
# Cria uma cópia da base para aplicar a limpeza
# A base original df permanece preservada
df_limpo = df.copy()

# Lista das colunas completamente vazias
colunas_vazias = [
    "UNNAMED_10",
    "UNNAMED_11",
    "UNNAMED_12",
    "UNNAMED_13"
]

# Remove as colunas 100% vazias
df_limpo = df_limpo.drop(columns=colunas_vazias)

# Substitui #N/D por uma categoria explícita
df_limpo["PR_CAT"] = df_limpo["PR_CAT"].replace(
    "#N/D",
    "NÃO INFORMADO"
)

# Confere o resultado
print("Quantidade de colunas após a remoção:", df_limpo.shape[1])
print(
    "Registros classificados como NÃO INFORMADO:",
    (df_limpo["PR_CAT"] == "NÃO INFORMADO").sum()
)
print(
    "Total de valores nulos na base:",
    df_limpo.isnull().sum().sum()
)
```

```python
# Registra a quantidade de linhas antes da remoção
registros_antes = len(df_limpo)

# Remove registros completamente duplicados
# reset_index reorganiza a numeração das linhas
df_limpo = df_limpo.drop_duplicates().reset_index(drop=True)

# Calcula os resultados
registros_depois = len(df_limpo)
duplicatas_removidas = registros_antes - registros_depois

print("Registros antes da limpeza:", registros_antes)
print("Duplicatas removidas:", duplicatas_removidas)
print("Registros após a limpeza:", registros_depois)
```

```python
# Conferência final da base limpa
print("CONFERÊNCIA DA BASE LIMPA")
print("-" * 40)

print("Número de registros:", df_limpo.shape[0])
print("Número de colunas:", df_limpo.shape[1])
print("Total de valores nulos:", df_limpo.isnull().sum().sum())
print("Total de duplicatas:", df_limpo.duplicated().sum())
print(
    "Valores #N/D restantes:",
    (df_limpo["PR_CAT"] == "#N/D").sum()
)
print(
    "Valores NÃO INFORMADO:",
    (df_limpo["PR_CAT"] == "NÃO INFORMADO").sum()
)

print("\nColunas da base limpa:")
print(df_limpo.columns.tolist())

print("\nTipos de dados:")
print(df_limpo.dtypes)
```

```python
#Sprint 4 – Estatística descritiva e agrupamentos
#Atividades:
#Identificar a coluna de número de filhos.
#Calcular média, mediana, desvio padrão e moda.
#Calcular os valores máximo e mínimo.
#Contar os registros válidos da coluna.
#Criar pelo menos dois agrupamentos com groupby() ou pivot_table().
#Comparar os resultados obtidos.
#Registrar uma breve interpretação dos resultados.
#Resultado esperado:
#Estatísticas obrigatórias calculadas e pelo menos dois agrupamentos apresentados e interpretados no notebook.

# Confere a coluna que representa o número de filhos dos clientes
print("COLUNA ANALISADA: CL_FHL")
print("Tipo de dado:", df_limpo["CL_FHL"].dtype)
print("Valores nulos:", df_limpo["CL_FHL"].isnull().sum())

print("\nDistribuição dos registros por número de filhos:")
print(
    df_limpo["CL_FHL"]
    .value_counts()
    .sort_index()
)
```

```python
# Seleciona a coluna referente ao número de filhos
filhos = df_limpo["CL_FHL"]

# Calcula as estatísticas descritivas exigidas
estatisticas_filhos = pd.DataFrame({
    "Estatística": [
        "Média",
        "Mediana",
        "Desvio padrão",
        "Moda",
        "Máximo",
        "Mínimo",
        "Contagem"
    ],
    "Resultado": [
        filhos.mean(),
        filhos.median(),
        filhos.std(),
        filhos.mode().iloc[0],
        filhos.max(),
        filhos.min(),
        filhos.count()
    ]
})

# Arredonda os resultados para duas casas decimais
estatisticas_filhos["Resultado"] = (
    estatisticas_filhos["Resultado"].round(2)
)

print("ESTATÍSTICAS DO NÚMERO DE FILHOS")
display(estatisticas_filhos)
```

### Estatísticas descritivas do número de filhos

- Média: 1,15 filho por registro.
- Mediana: 0 filho.
- Desvio padrão: 1,42.
- Moda: 0 filho.
- Máximo: 4 filhos.
- Mínimo: 0 filho.
- Contagem: 733.447 registros válidos.

A média indica que os registros correspondem, em média, a clientes com aproximadamente 1,15 filho. A mediana e a moda iguais a zero demonstram que os clientes sem filhos representam o grupo mais frequente da base.

Como a base registra compras e um mesmo cliente pode aparecer várias vezes, essas estatísticas representam os registros de compras, e não necessariamente a distribuição de clientes únicos.

```python
# Agrupa os registros de compras por gênero
compras_por_genero = (
    df_limpo
    .groupby("CL_GENERO")
    .size()
    .reset_index(name="QUANTIDADE_COMPRAS")
)

# Calcula a participação percentual de cada gênero
compras_por_genero["PERCENTUAL"] = (
    compras_por_genero["QUANTIDADE_COMPRAS"]
    / compras_por_genero["QUANTIDADE_COMPRAS"].sum()
    * 100
).round(2)

# Ordena do maior para o menor
compras_por_genero = compras_por_genero.sort_values(
    by="QUANTIDADE_COMPRAS",
    ascending=False
)

print("COMPRAS POR GÊNERO")
display(compras_por_genero)
```

### Agrupamento 1 — Compras por gênero

O gênero feminino apresentou 382.427 registros de compras, correspondendo a 52,14% da base limpa. O gênero masculino apresentou 351.020 registros, representando 47,86%.

Portanto, o gênero feminino possui a maior quantidade de compras registradas, com 31.407 registros a mais que o gênero masculino. Apesar dessa diferença, a distribuição das compras entre os gêneros é relativamente equilibrada.

```python
# Agrupa os registros de compras por categoria de produto
compras_por_categoria = (
    df_limpo
    .groupby("PR_CAT")
    .size()
    .reset_index(name="QUANTIDADE_COMPRAS")
)

# Calcula a participação percentual de cada categoria
compras_por_categoria["PERCENTUAL"] = (
    compras_por_categoria["QUANTIDADE_COMPRAS"]
    / compras_por_categoria["QUANTIDADE_COMPRAS"].sum()
    * 100
).round(2)

# Ordena da categoria com mais compras para a com menos compras
compras_por_categoria = compras_por_categoria.sort_values(
    by="QUANTIDADE_COMPRAS",
    ascending=False
)

print("COMPRAS POR CATEGORIA DE PRODUTO")
display(compras_por_categoria)
```

### Agrupamento 2 — Compras por categoria de produto

A categoria Alimentos apresentou 384.197 registros, correspondendo a 52,38% das compras da base limpa. Em seguida aparecem Higiene, com 18,77%, e Limpeza, com 17,54%.

As três principais categorias — Alimentos, Higiene e Limpeza — representam aproximadamente 88,69% dos registros de compras. Isso demonstra uma forte concentração das compras em produtos considerados essenciais.

A categoria Acessórios apresentou a menor participação entre as categorias identificadas, com 1,75%. Os registros classificados como NÃO INFORMADO representam apenas 0,44% da base limpa.
Os 3.228 registros classificados como NÃO INFORMADO representam apenas 0,44% da base limpa. Por se tratar de uma participação reduzida, esse grupo possui impacto limitado sobre os resultados gerais e não altera a ordem das categorias mais compradas. Ainda assim, sua existência foi registrada como uma limitação da qualidade dos dados.


```python
# Agrupa os registros de compras por segmento de cliente
compras_por_segmento = (
    df_limpo
    .groupby("CL_SEG")
    .size()
    .reset_index(name="QUANTIDADE_COMPRAS")
)

# Calcula a participação percentual de cada segmento
compras_por_segmento["PERCENTUAL"] = (
    compras_por_segmento["QUANTIDADE_COMPRAS"]
    / compras_por_segmento["QUANTIDADE_COMPRAS"].sum()
    * 100
).round(2)

# Ordena do segmento com mais compras para o menor
compras_por_segmento = compras_por_segmento.sort_values(
    by="QUANTIDADE_COMPRAS",
    ascending=False
)

print("COMPRAS POR SEGMENTO DE CLIENTE")
display(compras_por_segmento)
```

### Análise complementar — Compras por segmento de cliente

O segmento B apresentou 468.505 registros de compras, correspondendo a 63,88% da base limpa. O segmento C representou 27,99%, enquanto o segmento A apresentou participação de 8,14%.

Os segmentos B e C, juntos, concentram 91,87% das compras registradas. O resultado demonstra que as compras estão fortemente concentradas no segmento B.

A base não apresenta a descrição dos critérios utilizados para classificar os clientes nos segmentos A, B e C. Portanto, é possível analisar a participação de cada grupo, mas não atribuir características comerciais ou socioeconômicas a esses segmentos.

```python
# Cria o ranking dos produtos com mais registros de compras
produtos_mais_comprados = (
    df_limpo
    .groupby("PR_NOME")
    .size()
    .reset_index(name="QUANTIDADE_COMPRAS")
    .sort_values(
        by="QUANTIDADE_COMPRAS",
        ascending=False
    )
    .head(10)
)

# Calcula a participação de cada produto na base completa
produtos_mais_comprados["PERCENTUAL_DA_BASE"] = (
    produtos_mais_comprados["QUANTIDADE_COMPRAS"]
    / len(df_limpo)
    * 100
).round(2)

# Reorganiza a numeração do ranking
produtos_mais_comprados = produtos_mais_comprados.reset_index(drop=True)
produtos_mais_comprados.index = produtos_mais_comprados.index + 1

print("DEZ PRODUTOS MAIS COMPRADOS")
display(produtos_mais_comprados)
```

### Análise complementar — Produtos mais comprados

O Presunto Cozido liderou o ranking, com 12.719 registros de compras, correspondendo a 1,73% da base limpa. O segundo produto foi Sardinha, com 6.610 registros e participação de 0,90%.

Os dez produtos mais comprados representam aproximadamente 9,73% da base. Apesar da forte concentração das compras na categoria Alimentos, nenhum produto individual possui participação dominante. Isso indica que a demanda está distribuída entre diversos produtos do catálogo.

O Presunto Cozido se destaca dos demais produtos, pois possui quase o dobro dos registros apresentados pelo segundo colocado.

```python
# Conta a quantidade de compras registrada para cada produto
abc_produtos = (
    df_limpo
    .groupby("PR_NOME")
    .size()
    .reset_index(name="QUANTIDADE_COMPRAS")
    .sort_values(
        by="QUANTIDADE_COMPRAS",
        ascending=False
    )
    .reset_index(drop=True)
)

# Calcula a participação percentual de cada produto
abc_produtos["PERCENTUAL"] = (
    abc_produtos["QUANTIDADE_COMPRAS"]
    / abc_produtos["QUANTIDADE_COMPRAS"].sum()
    * 100
)

# Calcula o percentual acumulado
abc_produtos["PERCENTUAL_ACUMULADO"] = (
    abc_produtos["PERCENTUAL"].cumsum()
)

# Identifica quantos produtos são necessários para alcançar 80% das compras
quantidade_classe_a = (
    abc_produtos["PERCENTUAL_ACUMULADO"]
    .ge(80)
    .idxmax()
    + 1
)

# Identifica o limite de 95% para formar a classe B
limite_classe_b = (
    abc_produtos["PERCENTUAL_ACUMULADO"]
    .ge(95)
    .idxmax()
    + 1
)

# Classifica os produtos em A, B e C
abc_produtos["CLASSE_ABC"] = "C"
abc_produtos.loc[0:quantidade_classe_a - 1, "CLASSE_ABC"] = "A"
abc_produtos.loc[
    quantidade_classe_a:limite_classe_b - 1,
    "CLASSE_ABC"
] = "B"

# Calcula quanto a classe A representa no catálogo
total_produtos = len(abc_produtos)

percentual_catalogo_classe_a = (
    quantidade_classe_a / total_produtos * 100
)

# Cria um resumo das classes
resumo_abc = (
    abc_produtos
    .groupby("CLASSE_ABC", sort=False)
    .agg(
        QUANTIDADE_PRODUTOS=("PR_NOME", "count"),
        QUANTIDADE_COMPRAS=("QUANTIDADE_COMPRAS", "sum")
    )
    .reset_index()
)

resumo_abc["PERCENTUAL_PRODUTOS"] = (
    resumo_abc["QUANTIDADE_PRODUTOS"]
    / total_produtos
    * 100
).round(2)

resumo_abc["PERCENTUAL_COMPRAS"] = (
    resumo_abc["QUANTIDADE_COMPRAS"]
    / len(df_limpo)
    * 100
).round(2)

print("ANÁLISE ABC DOS PRODUTOS")
print("-" * 50)
print("Total de produtos:", total_produtos)
print("Produtos necessários para alcançar 80% das compras:", quantidade_classe_a)
print(
    "Percentual do catálogo necessário para alcançar 80%:",
    round(percentual_catalogo_classe_a, 2),
    "%"
)

print("\nRESUMO DAS CLASSES:")
display(resumo_abc)
```

A análise ABC foi realizada considerando a frequência dos registros de compras, pois a base não possui informações sobre quantidade vendida, preço ou faturamento.

A Classe A reuniu 91 produtos, equivalentes a 77,12% dos 118 produtos existentes na base. Esses produtos representam 80,59% dos registros de compras.

A Classe B possui 17 produtos, correspondentes a 14,41% do catálogo e 14,61% das compras. A Classe C contém 10 produtos, equivalentes a 8,47% do catálogo e 4,80% das compras.

O resultado não apresenta o comportamento clássico da regra de Pareto, em que aproximadamente 20% dos itens representariam 80% do movimento.
